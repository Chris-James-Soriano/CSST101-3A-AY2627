# 3. Using Comparison Operators.

```r
attendance <- 72
grade <- 70

print("Attendance is below 75.", attedance < 75)
print("Grade is belowe 75.", grade < 75)
print("Attendance is at least 75.", attendance >= 75)
print("Grade is at least", grade >= 75)
```

# 4. Using AND and OR.

```r
cat("The student is at risk.", grade < 75 && grade < 75, "\n")
cat("The student is at risk.", grade < 75 || grade < 75)
```

# 6. Multiple Conditions

```r
if(grade >= 90){
  print("Excellent")
}else if(grade >= 85){
  print("Verry Good")
}else if(grade >= 80){
  print("Good")
}else if(grade >= 75){
  print("Passed")
}else{
  print("Failed")
}
```

# 7. Creating a Function.

```r
classify_grade <- function(grades){
  if(grades >= 90){
  print("Excellent")
  }else if(grades >= 85){
    print("Verry Good")
  }else if(grades >= 80){
    print("Good")
  }else if(grades >= 75){
    print("Passed")
  }else{
    print("Failed")
  }
}
classify_grade(75)
classify_grade(88)
classify_grade(95)
classify_grade(70)
classify_grade(80)
```

# 8. Creating a Data Frame

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grades = c(90, 70, 85, 77, 65)
)

students
```

# 9. Adding a Classification Column.

```r
students$Performance <- ifelse(
  students$Grade >= 80,
  "Good",
  ifelse(
    students$Grade >= 75,
    "Passed",
    "Failed"
  )
)
students
```

# 10. Craeting a Simple Rule-Based System

```r
student_risk <- function(attendance, grades){
  if(attendance < 75 && grade < 75){
    print("High Risk.")
  }else if(attendance < 85 && grade < 80){
    print("Moderate Risk.")
  }else{
    print("Low Risk.")
  }
}
```

# 11. Apply the Rule to Multiple Students.

```r
students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)
students
```

# 12. Count the Results

```r
table(students$Risk)
```

# 13. Create a Visualization.

```r
risk_counts <- table(students$Risk)

barplot(
  risk_counts,
  min = "Student Risk Classification",
  xlab = "Risk level",
  ylab = "Numer of Students"
)
```

# 14. Add a Recommendations.

```r
recommendation <- function(risk){
  if(risk == "High Risk"){
    return("Immidiate academic intervention")
  }else if(risk == "Moderate Risk"){
    return("Academic monitoring")
  }else{
    return("Continue regular monitoring")
  }
}

students$Recommendation <- mapply(
  recommendation,
  students$Risk
)

students
```
